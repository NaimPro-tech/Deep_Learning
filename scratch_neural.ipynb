## Basic functions in NumPy

```python
import numpy as np
from typing import Callable, List, Dict
import matplotlib.pyplot as plt
import matplotlib
%matplotlib inline
```

```python
def square(x: np.ndarray)-> np.ndarray:
    '''square each element in the input ndarray'''
    return np.power(x,2)
```

```python
def leaky_relu(x: np.ndarray)-> np.ndarray:
    '''Applying leaky relu function to each element in ndarray'''
    return np.maximum(0.2*x, x)
```

```python
def deriv(func: Callable[[np.ndarray], np.ndarray],
            input_: np.ndarray,
            delta: float = 0.001)-> np.ndarray:
            '''Evaluates the derivative of function "func" to every in the input array'''
            return (func(input_+delta) - func(input_-delta))/(2*delta)
```

```python
Array_function = Callable[[np.ndarray],np.ndarray]
# A function takes ndarray as inputs/argument and produces an ndarray

Chain = List[Array_function]
# A chain is a list of functions
```

```python
def chain_length_2(chain: Chain, x: np.ndarray)->np.ndarray:
    '''Evaluates two function in a row , in a chain'''
    assert len(chain)==2, "Length of input chain must be 2"

    f1 = chain[0]
    f2 = chain[1]

    return f2(f1(x))
```

```python
def chain_length_3(chain: Chain, x: np.ndarray)-> np.ndarray:
    '''Evalutes three fuctions in a row, in a Chain'''
    assert len(chain)==3, "Length of input chain should be 3"

    f1 = chain[0]
    f2 = chain[1]
    f3 = chain[2]

    return f3(f2(f1))
```

```python
def sigmoid(x: np.ndarray)->np.ndarray:
    '''Apply the sigmoid function to each element in the input ndarray'''
    return(1/(1+np.exp(-x)))
```
