## Forward Pass

```python
import numpy as np
import matplotlib.pyplot as plt
from typing import Callable, List, Dict
%matplotlib inline
```

```python
def square(x: np.ndarray)->np.ndarray:
    return np.power(x, 2)
```

```python
def sigmoid(x: np.ndarray)->np.ndarray:
    return (1/(1+np.exp(-x)))
```

```python
def leaky_relu(x: np.ndarray)->np.ndarray:
    return np.maximum(0.2*x,x)
```

```python
Array_functions = Callable[[np.ndarray],np.ndarray]
Chain = List[Array_functions]
```

```python
def chain_length_2(chain: Chain, x: np.ndarray)->np.ndarray:
    '''Evaluates two functions in a row, in a chain'''
    assert len(chain)==2, 'Chain length must be 2'

    f1 = chain[0]
    f2 = chain[1]

    return f2(f1(x))
```

```python
def chain_length_3(chain: Chain, x: np.ndarray)->np.ndarray:
    assert len(chain)==3, "Chain length must be 3"

    f1=chain[0]
    f2=chain[1]
    f3=chain[2]

    return f3(f2(f1(x)))
```

## Lets test on practical example

```python
raw_data = np.arange(-2,3, dtype=float)
X = raw_data.reshape(-1,1)
```

```python
#create chain and run function
my_chain = [square, sigmoid]
predictions = chain_length_2(my_chain, X)
```

```python
print(f"Input Column vector: {X}")
print(f"Output prediction: {predictions}")
```
